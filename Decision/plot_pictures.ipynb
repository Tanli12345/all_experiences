```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib import font_manager as fm
import os

# ==================== 0. 环境与中文字体配置 ====================
def set_chinese_font():
    try:
        font_paths = ['/usr/share/fonts/truetype/wqy/wqy-microhei.ttc',
                      'C:/Windows/Fonts/simhei.ttf',
                      '/System/Library/Fonts/PingFang.ttc']
        for font_path in font_paths:
            if os.path.exists(font_path):
                fm.fontManager.addfont(font_path)
                plt.rcParams['font.family'] = fm.FontProperties(fname=font_path).get_name()
                plt.rcParams['axes.unicode_minus'] = False
                return True
    except: pass
    return False

set_chinese_font()

# ==================== 1. 绘制 MDP 学术原理图 ====================
def draw_mdp_matplotlib():
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.axis('off') # 关闭坐标轴

    # --- A. 绘制实体方框 (黑白学术风) ---
    # 1. 智能体 (Agent) 框
    agent_box = patches.FancyBboxPatch(
        (0.25, 0.65), 0.5, 0.2, 
        boxstyle="round,pad=0.02,rounding_size=0.05", 
        ec="black", fc="white", lw=2
    )
    ax.add_patch(agent_box)
    ax.text(0.5, 0.75, "智能体 (Agent)\n[Dueling DDQN 决策网络]", 
            ha='center', va='center', fontsize=15, fontweight='bold')

    # 2. 环境 (Environment) 框
    env_box = patches.FancyBboxPatch(
        (0.25, 0.15), 0.5, 0.2, 
        boxstyle="round,pad=0.02,rounding_size=0.05", 
        ec="black", fc="white", lw=2
    )
    ax.add_patch(env_box)
    ax.text(0.5, 0.25, "通信环境 (Environment)\n[卫星时频信道态势]", 
            ha='center', va='center', fontsize=15, fontweight='bold')

    # --- B. 绘制交互箭头与文本 ---
    # 1. 动作 (Action) - 从 Agent 向下指向 Env
    ax.annotate('', xy=(0.3, 0.35), xytext=(0.3, 0.65),
                arrowprops=dict(arrowstyle="-|>", lw=2.5, color="black", shrinkA=0, shrinkB=0))
    ax.text(0.28, 0.5, "动作 (Action) $A_t$\n[信道切换/保持指令]", 
            ha='right', va='center', fontsize=13)

    # 2. 状态 (State) - 从 Env 向上指向 Agent
    ax.annotate('', xy=(0.5, 0.65), xytext=(0.5, 0.35),
                arrowprops=dict(arrowstyle="-|>", lw=2.5, color="black", shrinkA=0, shrinkB=0))
    ax.text(0.52, 0.5, "状态 (State) $S_t$\n[态势演化特征矩阵]", 
            ha='left', va='center', fontsize=13)

    # 3. 奖励 (Reward) - 从 Env 向上指向 Agent
    ax.annotate('', xy=(0.7, 0.65), xytext=(0.7, 0.35),
                arrowprops=dict(arrowstyle="-|>", lw=2.5, color="black", shrinkA=0, shrinkB=0))
    ax.text(0.72, 0.5, "奖励 (Reward) $R_t$\n[抗干扰通信吞吐量增益]", 
            ha='left', va='center', fontsize=13)

    # --- C. 整体修饰与保存 ---
    plt.title("马尔可夫决策过程 (MDP) 闭环交互原理", fontsize=16, fontweight='bold', y=0.95)
    plt.tight_layout()
    plt.savefig("MDP_Block_Diagram_Academic.png", dpi=600, bbox_inches='tight')
    print("✅ MDP 黑白学术图已通过 Matplotlib 生成: MDP_Block_Diagram_Academic.png")
    plt.show()

if __name__ == "__main__":
    draw_mdp_matplotlib()
```
